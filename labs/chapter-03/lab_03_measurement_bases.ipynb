# Lab 03 — State Preparation and Measurement Bases
Prepare computational and superposition states, then compare Z- and X-basis probabilities.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator
import numpy as np

def prepare(*gates):
    qc = QuantumCircuit(1)
    for gate in gates:
        getattr(qc, gate)(0)
    return qc, Statevector.from_instruction(qc)

examples = {'|0>': prepare()[1], '|1>': prepare('x')[1], '|+>': prepare('h')[1], '|->': prepare('x','h')[1]}
for name, state in examples.items():
    print(name, state.data, state.probabilities())
```

## Step 2 — Measure in the X basis
A Hadamard before computational-basis readout changes the measurement basis.

```python
H = Operator(np.array([[1, 1], [1, -1]], dtype=complex) / np.sqrt(2))
plus = prepare('h')[1]
minus = prepare('x','h')[1]
plus_x = plus.evolve(H)
minus_x = minus.evolve(H)
print('|+> measured in X basis:', plus_x.probabilities())
print('|-> measured in X basis:', minus_x.probabilities())
assert np.allclose(plus_x.probabilities(), [1, 0])
assert np.allclose(minus_x.probabilities(), [0, 1])
```
