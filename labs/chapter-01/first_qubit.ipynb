## Step 2 — Build the circuit

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)
qc.h(0)
print(qc.draw())
```

## Step 3 — Inspect the exact ideal statevector
The target state is (|0⟩ + |1⟩)/√2.

```python
state = Statevector.from_instruction(qc)
print('Statevector:', state)
probabilities = state.probabilities_dict()
print('Exact probabilities:', probabilities)
assert abs(sum(probabilities.values()) - 1.0) < 1e-12
assert abs(probabilities.get('0', 0.0) - 0.5) < 1e-12
assert abs(probabilities.get('1', 0.0) - 0.5) < 1e-12
```
