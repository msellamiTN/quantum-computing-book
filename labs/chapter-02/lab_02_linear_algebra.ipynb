## Step 1 — Complex arithmetic
Compute conjugates and squared magnitudes.

```python
import numpy as np
z = 3 + 4j
print(z.conjugate(), abs(z)**2)
assert np.isclose(abs(z)**2, 25)
```

## Step 2 — Normalize a state vector

```python
v = np.array([1, 1j], dtype=complex)
v_norm = v / np.linalg.norm(v)
print(v_norm, np.linalg.norm(v_norm))
assert np.isclose(np.vdot(v_norm, v_norm), 1)
```

## Step 3 — Verify unitary gates

```python
X = np.array([[0, 1], [1, 0]], dtype=complex)
H = np.array([[1, 1], [1, -1]], dtype=complex) / np.sqrt(2)
I = np.eye(2)
assert np.allclose(X.conj().T @ X, I)
assert np.allclose(H.conj().T @ H, I)
print('X and H are unitary.')
```

## Step 4 — Compare with Qiskit

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
qc = QuantumCircuit(1)
qc.x(0)
sv = Statevector.from_instruction(qc)
print(qc.draw(), sv.data)
assert np.allclose(sv.data, X @ np.array([1, 0]))
```
