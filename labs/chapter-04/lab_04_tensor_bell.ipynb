## Step 1 — Tensor product by hand

```python
import numpy as np
plus = np.array([1,1], dtype=complex) / np.sqrt(2)
zero = np.array([1,0], dtype=complex)
product = np.kron(plus, zero)
print(product)
assert np.isclose(np.vdot(product, product), 1)
```

## Step 2 — Prepare the Bell state with Qiskit

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
bell = Statevector.from_instruction(qc)
print(qc.draw())
print('Amplitudes:', bell.data)
print('Probabilities:', bell.probabilities_dict())
assert np.allclose(bell.probabilities(), [0.5,0,0,0.5])
```

## Step 3 — Interpret
The state has amplitudes only on |00> and |11>. These outcomes occur with equal probability and are perfectly correlated. Qiskit uses qubit 0 as the least-significant bit; document bit order when reading bitstrings.
